```python
import pandas as pd
```

```python
import holidays
from datetime import date

def generate_yearly_profile(excel_path, sheet_name, year, output_path):
    """
    Converts a standard template load profile (Winter/Summer/Mid-Season) 
    into a continuous 1-column yearly time series.
    """
    
    # 1. Load the Excel file, skipping the first two header rows
    # We use header=None and assign our own clean column names based on the exact structure
    df = pd.read_excel(excel_path, sheet_name=sheet_name, skiprows=2, header=None)
    
    # Ensure we only take the first 10 columns if there's trailing empty data
    df = df.iloc[2:, 0:10]
    df.columns = [
        'Time',
        'Winter_Saturday', 'Winter_Sunday', 'Winter_Weekday',
        'Summer_Saturday', 'Summer_Sunday', 'Summer_Weekday',
        'Mid-Season_Saturday', 'Mid-Season_Sunday', 'Mid-Season_Weekday'
    ]
    df = df.dropna()

    # 2. Define Season Boundaries (Standard BDEW rules)
    def get_season(dt):
        # Format month and day as MMDD for easy comparison
        md = dt.month * 100 + dt.day
        
        if md < 321 or md > 1031:       # Nov 1 to Mar 20
            return 'Winter'
        elif 515 <= md <= 914:          # May 15 to Sep 14
            return 'Summer'
        else:                           # Mar 21 to May 14 & Sep 15 to Oct 31
            return 'Mid-Season'

    # 3. Initialize Holidays 
    # (Using Germany by default for BDEW profiles, adjust country code if needed)
    country_holidays = holidays.Germany(years=year)

    # 4. Define Day Type Logic
    def get_day_type(dt):
        if dt in country_holidays or dt.weekday() == 6:  # Sunday or Public Holiday
            return 'Sunday'
        elif dt.weekday() == 5:                          # Saturday
            return 'Saturday'
        else:
            return 'Weekday'                             # Monday - Friday

    # 5. Generate Daily Profiles
    yearly_data = []
    
    # Generate a date range for every day in the target year
    days_in_year = pd.date_range(start=f'{year}-01-01', end=f'{year}-12-31', freq='D')

    for day in days_in_year:
        season = get_season(day)
        day_type = get_day_type(day)
        col_name = f"{season}_{day_type}"
        
        # Extract the 96 values for this specific day and append to our master list
        daily_values = df[col_name].values
        yearly_data.extend(daily_values)

    # 6. Generate 15-minute timestamps for the output
    # This creates timestamps starting at 00:00 (Start of the interval)
    timestamps = pd.date_range(
        start=f'{year}-01-01 00:00', 
        periods=len(yearly_data), 
        freq='15min'
    )

    # 7. Format Output DataFrame
    out_df = pd.DataFrame({
        'Time': timestamps.strftime('%d.%m.%Y %H:%M'),
        'Sum [kWh]': yearly_data
    })
    
    # Set the index name to match your requested format
    out_df.index.name = 'Electricity.Timestep'

    # 8. Save to CSV
    # Using semicolon separator as requested in the target format
    out_df.to_csv(output_path, sep=';')
    print(f"Success! Generated {len(out_df)} rows for the year {year}. Saved to {output_path}")
```
